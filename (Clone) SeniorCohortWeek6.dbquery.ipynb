```sql
---SELECTING all rows and columns to decide what to include
SELECT *
FROM workspace.default.breath_of_the_wild_clean;

SELECT * 
FROM workspace.default.minecraft_data;

SELECT *
FROM workspace.default.mariokart_data;

---BUILDING OUR DATASET
DROP TABLE IF EXISTS tableau_data;

CREATE TABLE IF NOT EXISTS tableau_data
SELECT botw.PlayerID
    ,botw.Age
    ,botw.HoursPlayedPerWeek AS botw_hours_played
    ,botw.FavoriteWeapon
    ,botw.FavoriteRune
    ,botw.CombatSkillLevel
    ,botw.RupeesEarnedThisWeek
    ,botw.InventoryWeapons
    ,botw.ArcherySkillLevel
    ,mc.HoursPlayedPerWeek AS mc_hours_played
    ,mc.FavoriteMode
    ,mc.FavoriteMob
    ,mc.PvPSkillLevel
    ,mc.DiamondsFound
    ,mc.BiomePreference
    ,mc.TotalXP
    ,mc.PlayerLevel
    ,mk.FavoriteCharacter
    ,mk.FavoriteTrack
    ,mk.WinRate
    ,mk.SkillLevel
    ,mk.FavoriteKart
    ,mk.OnlineRank
    ,mk.Points
    ,mk.RacesCompleted
    ,mk.HoursPlayedPerWeek AS mk_hours_played

FROM workspace.default.breath_of_the_wild_clean AS botw

INNER JOIN workspace.default.minecraft_data AS mc
ON botw.PlayerID = mc.PlayerID

LEFT JOIN workspace.default.mariokart_data AS mk
ON botw.PlayerID = mk.PlayerID;

SELECT *
FROM workspace.default.tableau_data;
```
